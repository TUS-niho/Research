mp4の動画ファイルをwavの音声ファイルに変更

```python
import os
import moviepy.editor as mp

def batch_convert_mp4_to_wav(input_folder, output_folder):
    """
    指定されたフォルダ内の全てのMP4ファイルをWAVに変換し、
    別の指定されたフォルダに出力します。

    Args:
        input_folder (str): MP4ファイルが格納されているフォルダのパス
        output_folder (str): WAVファイルを出力するフォルダのパス
    """

    # 1. 出力フォルダが存在しない場合は作成する
    if not os.path.exists(output_folder):
        os.makedirs(output_folder)
        print(f"作成された出力フォルダ: {output_folder}")

    # 2. 入力フォルダ内の全てのファイル名を取得
    files = os.listdir(input_folder)

    # 3. MP4ファイルだけをリストアップ
    mp4_files = [f for f in files if f.endswith(('.mp4', '.MP4'))]

    if not mp4_files:
        print(f"'{input_folder}' にMP4ファイルが見つかりませんでした。")
        return

    print(f"{len(mp4_files)}個のMP4ファイルを変換します...")

    # 4. 各MP4ファイルをループ処理
    for mp4_file in mp4_files:
        # 入力ファイルのフルパスを作成
        input_path = os.path.join(input_folder, mp4_file)

        # 出力ファイルの名前を作成（拡張子を.wavに変更）
        file_name_without_ext = os.path.splitext(mp4_file)[0]
        output_filename = f"{file_name_without_ext}_audio.wav"
        output_path = os.path.join(output_folder, output_filename)

        print(f"変換中: {mp4_file} -> {output_filename}")

        try:
            # 動画ファイルを読み込む
            video = mp.VideoFileClip(input_path)

            # 音声をWAVファイルとして書き出す
            video.audio.write_audiofile(output_path)
            
            # リソースを解放
            video.close()

        except Exception as e:
            print(f"エラー: {mp4_file} の変換に失敗しました。詳細: {e}")

    print("すべての変換が完了しました。")


# --- 設定 ---
# 自分の環境に合わせてパスを書き換えてください

# MP4ファイルが入っているフォルダのパス
input_dir = r"D:\Douga_Niho\表情解析202404"

# WAVファイルを出力したいフォルダのパス
output_dir = r"D:\Douga_Niho\voice_data"
# ------------


# 関数を実行
batch_convert_mp4_to_wav(input_dir, output_dir)
```
