Confidenceが高いフレームだけを抽出

```python
import pandas as pd
from pathlib import Path
import os
import numpy as np

file_path = Path('../../../output_filtered_questions_copy')

file_list = list(file_path.glob('*.csv'))

for f in file_list:
    df = pd.read_csv(f)
    filename = os.path.basename(f)

    Confidence = df['Confidence'] < 0.95

    df.loc[Confidence, :] = np.nan

    output_folder = '../../../filtered_face_data_0.95'

    output_path = os.path.join(output_folder, f"{os.path.splitext(filename)[0]}_0.95.csv")

    print(df)
    
    df.to_csv(output_path, index=False, header=True)
```
