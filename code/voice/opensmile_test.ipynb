```python
import opensmile
import pandas as pd

# opensmile.Smileクラスの初期化時に、直接optionsを指定する
smile = opensmile.Smile(
    feature_set=opensmile.FeatureSet.eGeMAPSv02,
    feature_level=opensmile.FeatureLevel.LowLevelDescriptors,
)

# これでカスタム設定が適用されたsmileオブジェクトができます
result_df = smile.process_file('./voice_sample/asano.wav')
display(result_df.head())
```
